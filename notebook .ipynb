# 🚗 Car Agent AI

Welcome to **Car Agent AI**! 🎯

This project simulates a smart assistant that helps users diagnose basic car issues.
- The user asks a question about their car.
- The Car Agent suggests a possible service (brakes, tires, oil, headlights, etc.).

We also show some **statistics and visualizations** of questions asked by users.

```python
# Example: Simple Car Agent logic
def car_agent(question):
    question = question.lower()
    if "brake" in question:
        return "Brake service"
    elif "tire" in question:
        return "Tire service"
    elif "engine" in question or "overheat" in question:
        return "Cooling system flush"
    elif "oil" in question:
        return "Oil service"
    elif "headlight" in question:
        return "Headlight restoration"
    else:
        return "General inspection"

# Test example
print(car_agent("My brakes are making a squeaking noise when I stop"))
```

## 📊 User Questions Dataset
Below is a small dataset of user questions and the responses given by the Car Agent.

```python
import pandas as pd

data = {
    "Question": [
        "My brakes are squeaking",
        "The engine overheats quickly",
        "My tire is flat",
        "The headlights are too dim",
        "I need an oil change"
    ],
    "Response": [
        "Brake service",
        "Cooling system flush",
        "Tire service",
        "Headlight restoration",
        "Oil service"
    ],
    "Category": [
        "Brakes",
        "Engine",
        "Tires",
        "Headlights",
        "Oil"
    ]
}

df = pd.DataFrame(data)
df
```

## 📈 Visualizations

```python
import matplotlib.pyplot as plt

# Count categories
category_counts = df['Category'].value_counts()

# Bar chart
plt.figure(figsize=(6,4))
category_counts.plot(kind='bar')
plt.title("Number of Questions by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()

# Pie chart
plt.figure(figsize=(6,6))
category_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Distribution of Car Issues")
plt.ylabel("")
plt.show()
```
